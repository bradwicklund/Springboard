# Capstone Two: Preprocessing and Training Data Development

#### Clickstream Data for Online Shopping
#### Dataset: https://archive.ics.uci.edu/dataset/553/clickstream+data+for+online+shopping

In this notebook, we will be completing the three key steps of data preprocessing and training data development prior to building our machine learning model.

1. Create dummy or indicator features for categorical variables
2. Standardize the magnitude of numeric features
3. Split into testing and training datasets

```python
# Import packages
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
# Import data and convert date from object type
df = pd.read_csv('dataset/shopping_data_cleaned.csv')
df['date'] = pd.to_datetime(df['date'])
# Convert columns to object
cols_to_object = ['country', 
                        'page 1 (main category)', 
                        'page 2 (clothing model)', 
                        'colour', 
                        'location', 
                        'model photography']
df[cols_to_object] = df[cols_to_object].astype('object')
print(df.dtypes)
```

```python
# Convert price 2 to boolean
df['price 2_bool'] = (df['price 2'] == 1)
```

```python
# Drop extra columns
df = df.drop(columns = ['year', 'month', 'day', 'price 2'])
```

```python
# Inspect data frame
df.head()
```

## 1. Create dummy or indicator features from categorical variables

```python
# Select object type columns
dfo = df.select_dtypes(include=['object'])
df = pd.concat([df.drop(dfo, axis=1), pd.get_dummies(dfo)], axis=1)
df.head()
```

## 2. Standardize the magnitude of the numeric features

```python
# Initialize and fit-transform with StandardScaler
scaler = StandardScaler()
scaled_array = scaler.fit_transform(df[['price']])

# Extract the “price” column as a DataFrame, not a Series
price_df = df[['price']]

# Fit and transform 'price'
scaled_price = scaler.fit_transform(price_df)

# Put the scaled values back into DataFrame
df['price_scaled'] = scaled_price
```

## 3. Split into testing and training datasets

```python
# split data into test and training sets
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)
```
